```python
import duckdb 
import pandas as pd 
import os 
from datetime import datetime
```

```python
# Cria um banco de dados DuckDB persistente no arquivo dados_duckdb.db
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
arquivo = "z0019_2.csv"
data_ingestao = datetime.now()

# Gera DataFrame apartir do arquivo excel
df = pd.read_csv(f"C:/Users/Diego/Desktop/desenvolvimento-pos/myproject-pos-engenharia-dados-ia/landing/{arquivo}", sep=";")

# Adiciona coluna com nome do arquivo e uma coluna com data da ingestão vindos da variaveis 'arquivo' e 'data_ingestao'
df['nome_arquivo']  = arquivo
df['data_ingestao'] = data_ingestao

df.head()
```

### Tabela da camada Bronze

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS bronze_produtos(
        NATBR VARCHAR, 
        MAKTX VARCHAR, 
        WERKS VARCHAR,
        MAINS VARCHAR,	
        LABST VARCHAR,
        nome_arquivo VARCHAR,
        data_ingestao TIMESTAMP
    )
""")
```

```python
# Insert na tabela bronze_produtos apartir do DataFrame
con.execute("INSERT INTO bronze_produtos SELECT * FROM df ")
```

```python
# Validar insersão da tabela bronze_produtos
# O método fetch_df() reculpera o resultado da consulta SQL como um DataFrame para ser exibido no head()
resultado = con.execute("SELECT * FROM bronze_produtos").fetch_df()
resultado.head()
```
